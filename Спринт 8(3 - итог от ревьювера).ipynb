Логика merge (любого типа) для каждого значения ключа:

+ Найти все строки слева с этим id
+ Найти все строки справа с этим id
+ Соединить каждую с каждой (декартово произведение внутри группы)

Затем применить how — решить, какие комбинации оставить

Две категории:

| Категория | Доля продаж |
|-----------|-------------|
| A         | 90%         |
| B         | 10%         |

По долям кажется, что категория A — абсолютный лидер. А теперь добавим общее число продаж:

| Категория | Доля продаж | Всего продаж (шт.) | Продажи в штуках |
|-----------|-------------|--------------------|------------------|
| A         | 90%         | 10                 | 9                |
| B         | 10%         | 10 000             | 1 000            |

Абсолютные числа показывают, что категория B принесла 1 000 продаж против 9 у категории A, хотя её доля составляет всего 10%.

**Вывод:** 90% от 10 — это 9 продаж. 10% от 10 000 — это 1 000 продаж. Без абсолютных чисел легко переоценить значимость первой категории и недооценить вторую

```python
correlation_matrix = df[['rating', 'category', 'district', 'chain', 'seats',
                         'price', 'is_24_7']].phik_matrix()
# Выводим результат
print('Корреляционная матрица с коэффициентом phi_k для переменной rating')
correlation_matrix.loc[correlation_matrix.index != 'rating'][['rating']].sort_values(by='rating', ascending=False)
```

```python
plt.figure(figsize=(2, 6))

data_heatmap = correlation_matrix.loc[correlation_matrix.index != 'rating'][['rating']].sort_values(by='rating', ascending=False)
sns.heatmap(data_heatmap,
            annot=True, 
            fmt='.2f', 
            cmap='coolwarm', 
            linewidths=0.5, 
            cbar=False 
           )

plt.title('Тепловая карта коэффициента phi_k \n для данных rating')
plt.xlabel('Рейтинг заведения')

plt.show()
```

Из-за диспропорции данных (`sparse data`) связь между ценовым уровнем и рейтингом может оказаться ложной — на самом деле она отражает не влияние цены, а структуру самих категорий. Например, среди фастфуда почти нет высокой ценовой категории, а среди пабов — низкой. Поэтому пабы из-за своей значительной доли в высокой ценовой категории тянут рейтинг вверх именно в этом сегменте, а фастфуд — тянет вниз низкий сегмент, даже если внутри ценовых категорий связи с рейтингом нет
        
В чём проблема: 
        
по этой причине результаты твоего исследования могут быть искажены этой диспропорцией — отсутствием релевантного соотношения всех категорий цены во всех категориях заведений (['низкие', 'средние', 'выше среднего', 'высокие']). В аналитических терминах это называется `sparse data` или разрыв совместной встречаемости категорий.
        
Как с этим работать:

+ посчитай, сколько заведений каждой категории попало в выборку
+ если какой-то ценовой сегмент не представлен в категории (например, нет дорогих булочных) — делай выводы только в рамках доступных данных
+ для корректного сравнения используй нормировку       
        
Чтобы быстро увидеть диспропорцию, построй таблицу сопряженности (cross-tabulation) или её визуализацию — тепловую карту (heatmap), или группированную гистограмму
        
        pd.crosstab(df['category'], df['price'])
        # или с нормализацией
        pd.crosstab(df['category'], df['price'], normalize='index')
        
Если видишь нули или очень маленькие числа — корректно сравнивать ценовые уровни между такими категориями практически невозможно     
Что делать:

+ ограничить выводы только теми сочетаниями, где есть данные (наш случай)
+ объединить редкие категории (например, «выше среднего» + «высокие»)
+ либо вообще исключить проблемные категории из сравнения, сделав сноску в выводах 
+ признать, что взаимосвязь определить не получится из-за "разрыва совместной встречаемости категорий"        
        
Итог: в таких случаях самый честный вывод — не «связи нет», а «связь нельзя определить из-за диспропорции данных». Это нормально и профессионально        
        
p.s. не стоит забывать упомянуть в выводе и про значительное кол-во пропусков в колонке ['price']    
        
`sparse data` (разреженные данные) — это ситуация, когда в таблице сопряженности (crosstab) многие ячейки содержат ноль или очень маленькое количество наблюдений. Из-за этого невозможно делать надежные статистические выводы о связи между категориями — например, в твоей таблице у «столовой» всего 1 наблюдение с ценой «высокие», что делает любое сравнение с «рестораном» некорректным

Плюс: что делать с «Хинкальной» и «Чайхоной» в реальном проекте с неидеальными данными
        
Здесь нет единственного «правильного» ответа — есть выбор, и ты принимаешь его осознанно, документируя последствия. Вот основные стратегии:

1. Оставить как есть, но снять ограничения в выводах

Ты говоришь: «Мы не знаем, единая это сеть или нет, поэтому не делаем выводов, требующих этого знания». Например, не считаем количество «сетевых» заведений по имени, а говорим «заведения, имеющие одинаковое название и признак сети». Честно и профессионально.

2. Ввести порог «вхождения в сеть»

Например: считать сетью те названия, которые встречаются ≥ 3 или ≥ 5 раз в выборке. Это отсекает «Хинкальные»-одиночки (с категорией = кафе), но оставляет настоящие сети.

3. Использовать внешние данные

Подключить API (2ГИС, ФИАС, Дадата) и проверить, есть ли у «Хинкальной №1» и «Хинкальной №2» общий ИНН, юрлицо или сайт. Но это ресурсоёмко и не всегда возможно.

А если заказчик требует чёткого ответа? Ты ему говоришь:

«В данных есть заведения с одинаковым названием, но без подтверждения общего юрлица. Поэтому все выводы, связанные с сетями, сделаны с допущением, что одинаковое название = одна сеть. Погрешность в основном касается частотных названий вроде "Хинкальная". Для повышения точности потребуется дополнительная очистка данных или привлечение внешних источников.»

А функция ⬇ пригодится для быстрого знакомства с данными — несмотря на её объем она собрана из простых методов, большую часть из которых ты знаешь   

```python
# check
def inspect_data(df):
    
    """
        DataFrame со статистикой по каждой колонке:
    
        - rows_cnt: количество непустых значений
        - isna_perc: процент пропущенных значений
        - cnt_nunique: количество уникальных значений
        - dtype: тип данных колонки
        - memory: занимаемая память в МБ
        - mode: самое частотное значение (первая мода)
        
         (только для числовых колонок)            
        - min: минимальное значение  
        - mean: среднее значение  
        - std: стандартное отклонение              
        - median: медианное значение  
        - max: максимальное значение    
        - sum: сумма значений  
        - sum_abs: сумма абсолютных значений, для возможной оценки 
                между положительными и отрицательными значениями → sum_abs - sum 
        - neg_cnt: количество отрицательных значений в колонке
        
            
        Ограничения
        -----------
        Функция требует осторожности на больших данных (> 5 млн строк или > 500 колонок),
        особенно из-за операций nunique() и mode(), например:
            500 колонок при 1 млн строк (в сумме 500 млн ячеек)
            500 колонок + строковые/категориальные данные с высоким кардиналитетом (разнообразием значений)
        
        При недостатке памяти рекомендуется анализировать выборку колонок 
        >>> inspect_data(df.iloc[:, :20])  # только первые 20 колонок
        >>> # или
        >>> inspect_data(df[['col1', 'col2', 'col3']])
        
    """
    if df.empty:
        print("⚠️ DataFrame пуст")
        return pd.DataFrame()
    
    # Базовые метрики
    result = pd.DataFrame({
        'rows_cnt': df.count(),
        'isna_perc': (df.isna().mean() * 100).round(2),
        'cnt_nunique': df.nunique(),
        'dtype': df.dtypes.astype(str),
        'memory': (df.memory_usage(index=False) / 1024 / 1024).round(1),
        'mode': df.apply(lambda col: col.mode().iloc[0] if len(col.mode()) > 0 else pd.NA)        
    })
    
    # Статистики для числовых колонок
    numeric_cols = df.select_dtypes(include=['number', 'bool']).columns
    
    if len(numeric_cols) > 0:
        stats = df[numeric_cols].agg(['min', 'mean', 'std', 'median', 'max', 'sum']).round(2).T
        # Добавляем сумму абсолютных значений и кол-во отриц. значений в колонке
        stats['sum_abs'] = df[numeric_cols].abs().sum().round(2)      
        # считает КОЛИЧЕСТВО, т.к. (df[numeric_cols] < 0) выдает True or False (1 или 0)
        stats['neg_cnt'] = (df[numeric_cols] < 0).sum()         
        # объединяем базовые метрики с числовыми статистиками
        result = result.join(stats)
    
    # Сортируем по типу данных для удобства
    result = result.sort_values(['dtype'])
        
    return result        
```

```python
# check
inspect_data(df)
```
