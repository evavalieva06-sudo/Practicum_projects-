```python
def mean_sales_eu(row):
    if pd.isna(row['eu_sales']):
        sales = df[(df['platform'] == row['platform']) &
                          (df['year_of_release'] == row['year_of_release'])]
        return sales['eu_sales'].mean()
    else:
        return row['eu_sales']

    df['eu_sales'] = df.apply(mean_sales_eu, axis=1)
```

## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
def get_user_category(score):
    if pd.isna(score):
        return 'Без оценки'
    elif 0 <= score < 3:
        return 'Низкая оценка'
    elif 3 <= score < 8:
        return 'Средняя оценка'
    elif 8 <= score <= 10:
        return 'Высокая оценка'
    else:
        return 'Без оценки'

df_actual['category_user_score'] = df_actual['user_score'].apply(get_user_category)
user_counts = df_actual.groupby('category_user_score')['name'].count().sort_values(ascending=False)
print(user_counts)
```

- Разделите все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
def get_critic_category(score):
        if pd.isna(score):
                return 'Без оценки'
        elif 0 <= score < 30:
            return 'Низнкая оценка'
        elif 30 <= score < 80:
            return 'Средняя оценка'
        elif 80 <= score <= 100:
            return 'Высокая оценка'
        else:
            return 'Без оценки'
        
df_actual['category_critic_score'] = df_actual['critic_score'].apply(get_critic_category)
critic_counts = df_actual.groupby('category_critic_score')['name'].count().sort_values(ascending=False)
print(critic_counts)
```

- Выделите топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
top_platform = df_actual.groupby('platform')['name'].count().sort_values(ascending=False).head(7)
print(f'Топ-7 платформ: {top_platform}')
```
